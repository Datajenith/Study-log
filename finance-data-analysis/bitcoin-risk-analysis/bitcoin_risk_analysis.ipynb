```python
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
btc = yf.download("BTC-USD", start="2023-01-01", end="2026-05-19")

btc.head()
```

```python
btc["Close"].plot(figsize=(12,5))

plt.title("Bitcoin Closing Price")
plt.xlabel("Date")
plt.ylabel("Price")

plt.show()
```
