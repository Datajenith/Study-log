```
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf
```

```
btc = yf.download("BTC-USD", start="2023-01-01", end="2026-05-26")

btc.head()
```

```
btc["Return"] = btc["Close"].pct_change()

btc[["Close", "Return"]].head()
```

```
btc["Return"].plot(figsize=(12, 5))

plt.title("Bitcoin Daily Return")
plt.xlabel("Date")
plt.ylabel("Daily Return")

plt.show()
```

```
volatility = btc["Return"].std()

print("Bitcoin Volatility:", volatility)
```

```
btc["Cumulative Return"] = (1 + btc["Return"]).cumprod()

btc["Cumulative Return"].plot(figsize=(12, 5))

plt.title("Bitcoin Cumulative Return")
plt.xlabel("Date")
plt.ylabel("Cumulative Return")

plt.show()
```

```
rolling_max = btc["Cumulative Return"].cummax()

drawdown = btc["Cumulative Return"] / rolling_max - 1

mdd = drawdown.min()

print("Maximum Drawdown (MDD):", mdd)
```

## Analysis Summary

- Bitcoin showed high volatility during the analyzed period.
- Large price fluctuations resulted in significant drawdowns.
- The Maximum Drawdown (MDD) indicates that Bitcoin can experience substantial losses from its peak price.
- Return and volatility analysis are important for understanding investment risk.
